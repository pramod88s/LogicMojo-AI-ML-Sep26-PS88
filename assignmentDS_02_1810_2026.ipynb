```python
# Q29. Two models use the following input features:
#
# model_a_features = {"age", "salary", "tenure",
#                     "credit_score", "balance"}
#
# model_b_features = {"age", "salary", "tenure",
#                     "department", "region"}
#
# Find the common features and the features unique to each
# model.
# Calculate the Jaccard similarity between the feature sets.
#
# Formula:
# Jaccard Similarity = Size of Intersection / Size of Union
#
model_a_features = {"age", "salary", "tenure", "credit_score", "balance"}
model_b_features = {"age", "salary", "tenure", "department", "region"}

common_features = model_a_features & model_b_features
print("Common features:", common_features)
print("Features unique to Model A:", model_a_features - model_b_features)
print("Features unique to Model B:", model_b_features - model_a_features)

print("Jaccard Similarity = ", len(common_features) / len(model_a_features | model_b_features))
```

```python
# Q39. A dataset contains the following actual customer
# categories:
# actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]
# Create a dictionary that counts the frequency of each
# category.
# Then calculate the percentage of records belonging to
# each category.
# Example:
# Category A: 37.5%
# Print both the counts and percentages.
#
actual_categories = ["A", "B", "A", "C", "B", "A", "C", "B"]
actual_categories_freq = {}

for category in actual_categories:
    if category in actual_categories_freq:
        actual_categories_freq[category] += 1
    else:
        actual_categories_freq[category] = 1

print("Actual Categories frequency dict:", actual_categories_freq)        
print("")
for cat, freq in actual_categories_freq.items():
    print(f"For Category {cat}, records count is: {freq} and percentage of records is: {freq * 100 / len(actual_categories)}")
```

```python
# Q40. A data science team maintains two dictionaries
# containing model performance results:
# experiment_1 = {
# "Linear Regression": 0.82,
# "Decision Tree": 0.88,
# "Random Forest": 0.94
# }
# experiment_2 = {
# "Decision Tree": 0.90,
# "Random Forest": 0.96,
# "KNN": 0.85
# }
# Perform the following tasks:
# 1. Merge both dictionaries into a new dictionary.
# 2. If a model appears in both dictionaries, retain the
# higher accuracy score.
# 3. Print the final dictionary.
# 4. Identify the model with the highest accuracy.
# 5. Print all models whose final accuracy is greater than 0.90.
#
experiment_1 = {
"Linear Regression": 0.82,
"Decision Tree": 0.88,
"Random Forest": 0.94
}
experiment_2 = {
"Decision Tree": 0.90,
"Random Forest": 0.96,
"KNN": 0.85
}

merged = experiment_1.copy()
for model, accuracy in experiment_2.items():
    if model not in merged or accuracy > merged[model]:
        merged[model] = accuracy

print("Merged model performance:", merged)

highest_accuracy = 0;
highest_accurate_model = ""
for model, value in merged.items():
    if value > highest_accuracy:
        highest_accuracy = value
        highest_accurate_model = model

print(f"Model with highest accuracy is: {highest_accurate_model} and it's accuracy is: {highest_accuracy}")

for model, value in merged.items():
    if value > 0.90:
        print(f"Model {model} is having accuracy greater than 0.90 and is: {value}")
```
