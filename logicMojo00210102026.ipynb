```python
mark = []
i = 0
k = 0
print("Enter marks of 5 students")
while i < 5:
    m = int(input())
    i = i + 1
    if m < 0 or m > 100:
        print("Invalid marks entered")
        continue
    mark.append(m)    
    k = k + 1

for j in range(len(mark)):
    if mark[j] >= 90:
        print("Grade A")
    elif mark[j] >= 75 and mark[j] <= 89:
        print("Grade B")
    elif mark[j] >= 60 and mark[j] <= 74:
        print("Grade C")
    elif mark[j] >= 40 and mark[j] <= 59:
        print("Pass")
    elif mark[j] == 0:
        break
    elif mark[j] < 0:
        pass
    else:
        print("f")

print("Valid marks collected: ", k)        
```

```python
print("Calculate BMI...")
while True:
    print("Enter weight in kg:")
    weight = int(input())
    if weight == 0:
        break;
    
    print("Enter height in metres:")
    height = int(input())

    if weight <= 0 or height <= 0:
        print("Invalid input, please enter values greater than 0.")
        continue
    bmi = float(weight / (height*height))
    print("BMI: ", bmi)
    if bmi < 18.5:
        print("Underweight")
    elif bmi >= 18.5 and bmi < 25:
        print("Normal weight")
        print("Congrats...")
    elif bmi >= 25 and bmi < 30:
        print("Overweight")
    elif bmi >= 30:
        print("Obesity range")

        

print("Exiting...")
for j in range(1, 4):
    print(j)
```
