```python
==========================================
# PYTHON TUPLE: ALL METHODS AND OPERATIONS
# ==========================================

# 1. Creating a tuple
numbers = (10, 20, 30, 20, 40, 50, 20)
print(numbers)

# 2. Checking the type
print(type(numbers))

# 3. Finding the length
print(len(numbers))

# 4. Accessing elements using positive indexing
print(numbers[0])
print(numbers[2])

# 5. Accessing elements using negative indexing
print(numbers[-1])
print(numbers[-2])

# 6. Slicing
print(numbers[1:4])
print(numbers[:4])
print(numbers[2:])
print(numbers[::2])
print(numbers[::-1])

# 7. Counting occurrences using count()
print(numbers.count(20))
print(numbers.count(50))

# 8. Finding the index using index()
print(numbers.index(30))
print(numbers.index(20))
print(numbers.index(20, 2))  # Search starting from index 2

# 9. Checking membership using in
print(30 in numbers)
print(100 in numbers)

# 10. Checking membership using not in
print(100 not in numbers)
# 11. Finding the minimum
print(min(numbers))

# 12. Finding the maximum
print(max(numbers))

# 13. Calculating the sum
print(sum(numbers))

# 14. Concatenating two tuples
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
print(tuple1 + tuple2)

# 15. Repeating a tuple
print(tuple1 * 3)

# 16. Converting a tuple into a list
numbers_list = list(numbers)
print(numbers_list)

# 17. Converting a list into a tuple
new_tuple = tuple([100, 200, 300])
print(new_tuple)

# 18. Tuple unpacking
student = ("Aman", 21, 85)
name, age, marks = student
print(name)
print(age)
print(marks)

# 19. Extended tuple unpacking
values = (10, 20, 30, 40, 50)
first, *middle, last = values
print(first)
print(middle)
print(last)

# 20. Creating a single-element tuple
single_element = (100,)
print(single_element)
print(type(single_element))
# 21. Creating an empty tuple
empty_tuple = ()
print(empty_tuple)

# 22. Sorting a tuple using sorted()
# sorted() returns a LIST, not a tuple
unsorted_tuple = (50, 10, 40, 20, 30)
print(sorted(unsorted_tuple))
print(tuple(sorted(unsorted_tuple)))

# 23. Reversing a tuple using slicing
print(numbers[::-1])

# 24. Iterating through a tuple
for item in tuple1:
    print(item)

# 25. Enumerating tuple elements with their indexes
for index, value in enumerate(tuple1):
    print(index, value)

# 26. Finding the number of unique elements
print(len(set(numbers)))

# 27. Checking whether two tuples are equal
print((1, 2, 3) == (1, 2, 3))

# 28. Comparing tuples
print((1, 2, 3) < (1, 2, 4))

# 29. Getting the first and last elements
print(numbers[0], numbers[-1])

# 30. Demonstrating tuple immutability
# numbers[0] = 100  # Uncomment to see TypeError
# 31. Demonstrating that tuples can contain mutable objects
data = (10, 20, [30, 40])
data[2].append(50)
print(data)

# 32. Deleting a tuple variable
# del numbers  # Uncomment to delete the entire variable

# 33. Viewing the available tuple methods
print(dir(numbers))
```

```python
# 1. Creating a set
numbers = {10, 20, 30, 40, 50}
print(numbers)

# 2. Checking the type
print(type(numbers))

# 3. Creating an empty set
empty_set = set()
print(empty_set)

# 4. Creating a set from a list
numbers_from_list = set([10, 20, 20, 30, 30, 40])
print(numbers_from_list)

# 5. Finding the length
print(len(numbers))

# 6. Adding one element using add()
numbers.add(60)
print(numbers)

# 7. Adding an existing element
numbers.add(20)
print(numbers)

# 8. Adding multiple elements using update()
numbers.update([70, 80, 90])
print(numbers)

# 9. Updating from another set
numbers.update({100, 110})
print(numbers)

# 10. Removing an element using remove()
numbers.remove(20)
print(numbers)

# 11. Removing an element using discard()
numbers.discard(30)
print(numbers)
# 12. Discarding a nonexistent element
numbers.discard(999)
print(numbers)

# 13. Removing an arbitrary element using pop()
removed_value = numbers.pop()
print("Removed:", removed_value)
print(numbers)

# 14. Checking membership using in
print(40 in numbers)

# 15. Checking membership using not in
print(999 not in numbers)

# 16. Creating two sets for set operations
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# 17. Union using the | operator
print(A | B)

# 18. Union using union()
print(A.union(B))

# 19. Intersection using the & operator
print(A & B)

# 20. Intersection using intersection()
print(A.intersection(B))

# 21. Difference using the - operator
print(A - B)

# 22. Difference using difference()
print(A.difference(B))

# 23. Reverse difference
print(B - A)
# 24. Symmetric difference using ^ 
print(A ^ B)

# 25. Symmetric difference using symmetric_difference()
print(A.symmetric_difference(B))

# 26. Checking whether A is a subset of B
print(A.issubset(B))

# 27. Checking whether one set is a subset of another
print({1, 2}.issubset(A))

# 28. Checking whether A is a superset of another set
print(A.issuperset({1, 2}))

# 29. Checking whether two sets are disjoint
print({1, 2}.isdisjoint({3, 4}))

# 30. Updating a set using intersection_update()
C = {1, 2, 3, 4}
D = {3, 4, 5, 6}
C.intersection_update(D)
print(C)

# 31. Updating a set using difference_update()
C = {1, 2, 3, 4}
D = {3, 4, 5, 6}
C.difference_update(D)
print(C)

# 32. Updating a set using symmetric_difference_update()
C = {1, 2, 3, 4}
D = {3, 4, 5, 6}
C.symmetric_difference_update(D)
print(C)

# 33. Updating a set using |=
C = {1, 2}
C |= {3, 4}
print(C)

# 34. Updating a set using &=
C = {1, 2, 3}
C &= {2, 3, 4}
print(C)
# 35. Updating a set using -=
C = {1, 2, 3}
C -= {2}
print(C)

# 36. Updating a set using ^=
C = {1, 2, 3}
C ^= {3, 4}
print(C)

# 37. Copying a set using copy()
original = {10, 20, 30}
copied = original.copy()
copied.add(40)
print(original)
print(copied)

# 38. Converting a set into a list
numbers_list = list(A)
print(numbers_list)

# 39. Converting a set into a tuple
numbers_tuple = tuple(A)
print(numbers_tuple)

# 40. Removing duplicates from a list using set()
duplicate_list = [1, 2, 2, 3, 3, 4, 4, 5]
unique_values = set(duplicate_list)
print(unique_values)

# 41. Iterating through a set
for item in A:
    print(item)

# 42. Finding minimum and maximum
print(min(A))
print(max(A))

# 43. Calculating the sum
print(sum(A))

# 44. Comparing sets for equality
print({1, 2, 3} == {3, 2, 1})

# 45. Clearing all elements using clear()
C = {10, 20, 30}
C.clear()
print(C)

# 46. Viewing available set methods
print(dir(A))
```
