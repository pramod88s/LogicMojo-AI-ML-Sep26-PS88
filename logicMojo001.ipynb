```python
# Raw customer information
customer_name = "  LOGIC MOJO  "
email = " LOGIC.MOJO@GMAIL.COM "
purchase_amount = "2500.50"
customer_id = "CUST1025"
```

```python
# Step 1: Clean text
clean_name = customer_name.strip().title()
clean_email = email.strip().lower()

print("customer name: " + clean_name)
print("customer email: " + clean_email)
```

```python
# Step 2: Convert the amount to a number
purchase_amount = float(purchase_amount)
print(f"purchase amount: {purchase_amount}")
```

```python
# Step 3: Calculate tax and total
tax_rate = 0.18
tax_amount = purchase_amount * tax_rate
total_amount = purchase_amount + tax_amount
print(f"tax amount: {tax_amount}")
print(f"tax amount: {total_amount}")
```

```python
# Step 4: Extract information from the ID
id_prefix = customer_id[:4]
id_number = customer_id[4:]
print("customer id prefix: " + id_prefix)
print("customer id number: " + id_number)
print("customer id reverse: " + customer_id[::-1])
```

```python
# Step 5: Display the customer report
print("CUSTOMER REPORT")
print("-" * 30)
print(f"Customer name: {clean_name}")
print(f"Email: {clean_email}")
print(f"ID prefix: {id_prefix}")
print(f"ID number: {id_number}")
print(f"Purchase amount: ₹{purchase_amount:.2f}")
print(f"Tax amount: ₹{tax_amount:.2f}")
print(f"Total amount: ₹{total_amount:.2f}")
```
