```python
# Q1. Age Validation
# Take age as input from the user. If the input is empty, raise a ValueError. If the age is negative, raise another ValueError. Use try-except to handle the errors. If the age is 18 or above, print "Eligible to vote"; otherwise, print "Not eligible to vote".
#
age = input("Enter age:")

try:
    if not age:
        raise ValueError("Age cannot be empty")
    
    age_num = int(age)    
    
    if age_num < 0:
        raise ValueError("Age cannot be negative")
    elif age_num >= 18:
        print("Eligible to vote")
    else:
        print("Not eligible to vote")
        
except ValueError as e:
    print(e)
```

```python
# Q2. Positive Number Checker
# Ask the user to enter a number. Handle invalid numeric input using try-except. If the number is zero or negative, raise a ValueError. 
# If it is positive, check whether it is even or odd and display the result.
# 
try:
    num = int(input("Enter a number:"))
    
    if num <= 0:
        raise ValueError("Number should be greater than 0.")
    elif num % 2 == 0:
        print(f"Number {num} is even")
    elif num % 2 != 0:
        print(f"Number {num} is odd")
        
except ValueError as e:
    print(e)
```

```python
# Q3. Username Validation
# Take a username as input. If the username is empty or contains fewer than 5 characters, raise a ValueError. 
# Use try-except to handle the exception. If the username is valid, print "Username accepted".
#
username = input("Enter username:")

try:
    if not username or len(username) < 5:
        raise ValueError("Username is invalid. It should have minimum 5 characters.")
    else:
        print("Username accepted")
        
except ValueError as e:
    print(e)
```

```python
# Q4. Division Calculator
# Take two numbers as input. Handle invalid input using try-except. If the second number is zero, raise a ZeroDivisionError. 
# Otherwise, divide the first number by the second and print the result. Use else to display the result only when no exception occurs.
#


try:
    num1 = int(input("Enter num1:"))
    num2 = int(input("Enter num2:"))

    if num2 == 0:
        raise ZeroDivisionError("Error: Division by 0 is not allowed.")
    else:
        print(f"Divison of {num1} by {num2}: {num1 / num2}")
except ZeroDivisionError as e:
    print(e)
```

```python
# Q5. Student Marks Validation
# Take marks as input. If the input is not a valid integer, handle the exception. If marks are less than 0 or greater than 100, raise a ValueError. 
# Otherwise, use if-else to print "Pass" if marks are at least 40 and "Fail" otherwise.
#
try:
    marks = int(input("Enter marks:"))

    if marks < 0 or marks > 100:
        raise ValueError("Marks should be between 0 and 100.")
    elif marks >= 40:
        print("Pass")
    else:
        print("Fail")
except ValueError as e:
    print("Error:", e)
```

```python
# Q7. Write Student Records
# Use "w" mode to create or overwrite students.txt. 
# Take the details of three students and write each student's name and marks on a separate line. Handle any appropriate input errors using try-except.
#
try:
    file = open("students.txt", "w")

    for i in range(1, 4):
        print(f"Student {i} details:")
        name = input("Enter name:")
        marks = int(input("Enter marks:"))

        if i > 1:
            file.write("\n")
        file.write(f"Name: {name}, Marks: {marks}")
except ValueError as e:
    print("Error:", e)
finally:
    file.close()
```

```python
# Q8. Append Student Records
# Use "a" mode to add a new student's name and marks to the existing students.txt file. 
# If the marks are below 0 or above 100, raise a ValueError. Handle the exception without adding invalid records to the file.
#
try:
    file = open("students.txt", "a")

    print(f"Student details:")
    name = input("Enter name:")
    marks = int(input("Enter marks:"))

    if marks < 0 or marks > 100:
        raise ValueError("Marks should be greater than 0 and less than 100.")
    
    file.write("\n")
    file.write(f"Name: {name}, Marks: {marks}")
except ValueError as e:
    print("Error:", e)
finally:
    file.close()
```

```python
# Q9. Read a File Safely
# Ask the user to enter a filename. Open the file using "r" mode and display its contents. 
# Handle FileNotFoundError if the file does not exist. If the file is empty, print "File is empty".
#

file_name = input("Enter file name:")

try:
    file = open(file_name, "r")

    content = file.read()
    if not content:
        print("File is empty.")
    else:
        print("File content:")
        print(content)
except FileNotFoundError as e:
    print("Error:", e)
finally:
    if file:
        file.close()
```

```python
# Q10. Number Storage in a File
# Take five numbers from the user and write them to numbers.txt using "w" mode. If a user enters an invalid number, handle the exception. 
# If a number is negative, raise a ValueError. Store only valid, non-negative numbers in the file.
#
nums = []
count = 5
print(f"Enter {count} numbers:")

for i in range(1, count + 1):
    try:
        num = int(input(f"Enter number {i}:"))
    
        if num < 0:
            raise ValueError("Number should be non-negative.")
    
        nums.append(num)
    except ValueError as e:
        print("Error:", e)
    
try:
    file = open("numbers.txt", "w")

    for num in nums:
        file.write(f"{num}\n")

except FileNotFoundError as e:
    print("Error:", e)
finally:
   if file:
        file.close()
```

```python
# Q11. Search for a Student
# Read students.txt using "r" mode. Ask the user to enter a student name. 
# If the name exists in the file, print "Student found"; otherwise, print "Student not found". Handle FileNotFoundError if the file does not exist.
#

try:
    file = open("students.txt", "r")

    student_name = input("Enter student name to search:")
    content = file.read()
    if student_name in content:
        print("Student found")
    else:
        print("Student not found")
except FileNotFoundError as e:
    print("Error:", e)
finally:
   if file:
        file.close()
```

```python
# Q12. File Creation with Validation
# Ask the user to enter a filename. If the filename is empty, raise a ValueError. 
# Otherwise, use "x" mode to create the file. Handle both ValueError and FileExistsError appropriately.
#

file_name = input("Enter file name:")

try:
    if not file_name:
        raise ValueError(f"Invalid file name entered.")
    else:
        file = open(file_name, "x") 
        print(f"File with name {file_name} created successfully.")
except ValueError as ve:
    print("Error:", ve)
except FileExistsError as fe:
    print("Error:", fe)
finally:
    if file:
        file.close()
```

```python
# Q13. Append Only Passing Students
# Take a student's name and marks as input. If marks are outside the range 0–100, raise a ValueError.
# If marks are 40 or above, append the student's details to passed_students.txt using "a" mode. Otherwise, print "Student failed; record not added".
#
try:
    print(f"Student details:")
    name = input("Enter name:")
    marks = int(input("Enter marks:"))

    if marks < 0 or marks > 100:
        raise ValueError("Marks should be between 0 and 100.")
    elif marks >= 40:
        file = open("passed_students.txt", "a")
        file.write("\n")
        file.write(f"Name: {name}, Marks: {marks}")
    else:
        print("Student failed; record not added")
except ValueError as e:
    print("Error:", e)
finally:
    if file:
        file.close()
```

```python
# Q14. Count Lines in a File
# Ask the user for a filename and read it using "r" mode. Count the number of non-empty lines in the file. Handle FileNotFoundError. 
# If the file contains no non-empty lines, print "No records found".
#
file_name = input("Enter file name:")

try:
    file = open(file_name, "r") 

    non_empty_lines = sum(1 for line in file if line.strip())

    if non_empty_lines == 0:
        print("No records found")
    else:
        print("Total lines in file are:", non_empty_lines)
        
except FileNotFoundError as fe:
    print("Error:", fe)
finally:
    if file:
        file.close()
```
