```python
# Take 10 actual values and 10 predicted values, one pair at a time. Calculate the mean squared error (MSE) using a loop

total_sq_error = 0.0
count = 10
print(f"Enter {count} pairs of actual and predicted values")

for i in range(1, count + 1):
    actual = float(input(f"Enter actual value of pair {i} out of {count} pairs:"))
    predicted = float(input(f"Enter predicted value of pair {i} out of {count} pairs:"))

    error = actual - predicted
    total_sq_error += error ** 2

print(f"Mean Squared Error:", total_sq_error / count)
```

```python
# Write a function that accepts any number of model scores using *args and returns the highest score without using lists or built-in maximum functions.

def find_highest_score(*scores):
    if not scores:
        return None
    
    max = scores[0]
    for score in scores:
        if score > max:
            max = score

    return max

print("Highest score:", find_highest_score(0.98, 0.95, 0.99, 0.93, 0.80, 0.89, 0.99))
print("Highest score:", find_highest_score())
print("Highest score:", find_highest_score(-0.98, -0.95, -0.99, -0.93, -0.80, -0.89, -0.99))
```

```python
# Write a function that accepts any number of validation scores using *args and returns their average. Then use it to compare two model averages with a conditional statement

def compute_avg_validation_score(*scores):
    if not scores:
        return 0.0
    
    total_score = 0.0
    for score in scores:
        total_score += score
        
    return total_score / len(scores)

model_a_avg_score = compute_avg_validation_score(0.97, 0.95, 0.90, 0.93, 0.80, 0.89, 0.99)
model_b_avg_score = compute_avg_validation_score(0.98, 0.95, 0.99, 0.93, 0.80, 0.89, 0.99)

print(f"Model a average score: {model_a_avg_score:.2f}")
print(f"Model b average score: {model_b_avg_score:.2f}")

if model_a_avg_score > model_b_avg_score:
    print("Model a is better on average score")
elif model_b_avg_score > model_a_avg_score:
    print("Model b is better on average score")
else:
    print("Model a and Model b are equal on average score")
```

```python
# Write a decorator that prints "Training started" before a function executes and "Training finished" after it executes. Apply it to a function that prints the current epoch.

def decorate_epoch(func):
    def wrapper(*args, **kwargs):
        print("Training started")
        func(*args, **kwargs)
        print("Training finished")
    return wrapper    

@decorate_epoch
def print_current_epoch(epoch):
    print("Current Epoch: ", epoch)

print_current_epoch(10)
```

```python
# Write a decorator that accepts a function, executes it, and returns its result. Demonstrate it using a function that calculates model accuracy.

def decorate_accuracy_computation(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)

    return wrapper    

@decorate_accuracy_computation
def compute_model_accuracy(correct_predictions, total_predictions):
    return (correct_predictions * 100) / total_predictions

print(f"Model accuracy: {compute_model_accuracy(90, 100):.2f}%")
```

```python
# Build a mini training simulation using functions, a decorator, a loop, conditions, break, and a lambda function. Print the loss for each epoch, stop when loss reaches 0.10 or lower, and return the final loss. Use only the Python concepts covered in class.
get_loss_reduction = lambda loss, learning_rate: loss * learning_rate

def decorate_training(func):
    def wrapper(*args, **kwargs):
        print("Training Simulation Started ...")
        result = func(*args, **kwargs)
        print("Training Simulation Stopped ...")
        return result
    return wrapper        

@decorate_training
def train_model(starting_loss, learning_rate, epochs):
    loss = starting_loss
    loss_threshold = 0.10
    
    for epoch in range(1, epochs + 1):
        print(f"Epoch {epoch}: Current loss = {loss:.2f}")

        if loss <= loss_threshold:
            print(f"Target reached, so stopping at epoch {epoch}.")
            break

        reduction = get_loss_reduction(loss, learning_rate)
        loss -= reduction

    return loss

final_loss = train_model(starting_loss = 1.0, learning_rate = 0.15, epochs = 20)
print(f"Final loss during training: {final_loss:.2f}")
```
